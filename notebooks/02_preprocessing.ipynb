```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib as plt


df = pd.read_csv("../data/treated/heart_treated.csv")
df.head()
```

### Standardizing the columns names

```python
df = df.rename(columns=str.lower)
```

```python
df.dtypes
```

### Categorical variables encoding

```python
df["sex"] = df["sex"].map({"M": 0, "F": 1})
```

```python
df["exerciseangina"] = df["exerciseangina"].map({"Y": 1, "N": 0})
```

```python
df = pd.get_dummies(df, columns=["restingecg", "st_slope", "oldpeak_risk", "chestpaintype"], drop_first=True)
```

```python
df
```

```python
df = df.rename(columns=str.lower)
```

```python
df
```

### Checking for missing values

```python
df.isna().sum()
```

```python
df["cholesterol"] = df["cholesterol"].fillna(df["cholesterol"].median())
```

### Scaling

```python
X = df.drop("heartdisease", axis=1)
y = df["heartdisease"]

X_numeric = ["age", "restingbp", "cholesterol", "maxhr", "oldpeak", "hr_ratio"]
```

```python
X_cat = [col for col in X.columns if col not in X_numeric]
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = pd.DataFrame(
    scaler.fit_transform(X[X_numeric]), 
    columns = X_numeric,
    index = X.index
)
```

```python
### Concatenating the scaled variables (X_scaled) with the categorical ones (X_cat)
X_final = pd.concat([X_scaled, X[X_cat]], axis=1)
```

### Exporting final dataset

```python
df_final = pd.concat([X_final, y], axis=1)
```

```python
df_final_csv = df_final.to_csv("../data/treated/df_final.csv", index=False)
```
